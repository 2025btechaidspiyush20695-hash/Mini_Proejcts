```python
import numpy as np
number_of_students = 5
number_of_toses = 5
coin_toses = np.random.binomial(n=number_of_toses,p=0.5,size=number_of_students)
for num_heads in range(number_of_toses + 1):
    frequency = np.count_nonzero(coin_toses == num_heads)
    rf = frequency/number_of_students
    print(f"RF({num_heads}) = {rf}")
```
