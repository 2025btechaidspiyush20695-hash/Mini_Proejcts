```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)
n = int(input("Enter a number to find its factorial: "))
result = factorial(n)
print(f"The factorial of {n} is {result}")
```

```python
def find_greatest_number(my_array):
    if not my_array:
        return None
    greatest = my_array[0]
    for num in my_array:
        if num > greatest:
            greatest = num
    return greatest 

n = int(input("Array mein kitne elements daalne hain? (n): "))
arr = []

for i in range(n):
    num = int(input(f"Element {i+1} enter karein: "))
    arr.append(num)
result = find_greatest_number(arr)
print(f"\nAapki array hai: {arr}")
print(f"Array ka sabse bada (greatest) number hai: {result}")
```

```python
def celsius_to_fahrenheit(celsius):
    return (celsius * 9/5) + 32

C = float(input("Celsius mein temperature enter karein: "))
F = celsius_to_fahrenheit(C)
print(f"{C}°C = {F}°F")
```
