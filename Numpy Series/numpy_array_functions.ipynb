```python
import numpy as np
```

```python
A1 = np.random.random((3,3))
A1 = np.round(A1 * 100)
A1
```

```python
print(np.max(A1))
print(np.min(A1))
print(np.sum(A1))
print(np.prod(A1))
print(np.max(A1,axis=0))
print(np.max(A1,axis=1))
```

```python
#statastics
print(np.mean(A1))
print(np.median(A1))
print(np.std(A1))
print(np.var(A1))
```

```python
#triganometric functions
print(np.sin(A1))
```

```python
A2 = np.arange(12).reshape(3,4)
A3 = np.arange(12,24).reshape(4,3)

print(A2)
print("\n")
print(A3)
```

```python
#dot product
print(np.dot(A2,A3))
```

```python
#log 
print(np.log(A2))
#exponential
print("\n")
print(np.exp(A1))
```

```python
print(np.round(np.random.random((2,3))*100))
print("\n")
print(np.floor(np.random.random((2,3))*100))
print("\n")
print(np.ceil(np.random.random((2,3))*100))
```
